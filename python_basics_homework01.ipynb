````python
# TODO: write your code here
word_list = ["book", "school", "ship", "desk", "eye"]
meaning_list = ["책", "학교", "배", "책상", "눈"]

def check_answer(user_answer, correct_answer) : return user_answer.strip().lower() == correct_answer.strip().lower()

score = 0
total_questions = 3

for i in range(total_questions):
    index = random.randint(0, len(word_list) - 1)

    word = word_list[index]
    correct_meaning = meaning_list[index]
    user_answer = input(f"What is the meaning of {word}? ")

    if check_answer(user_answer, correct_meaning):
        print("Pass")
        score += 1

print(f"Final score: {score}/{total_questions}")
````
